```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler   
```

```python
data={"Age":[10,20,30,40,55]}
df=pd.DataFrame(data)
print(df)
```

```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[["Age"]])

df_scaled = pd.DataFrame(
    scaled,
    columns=["Age_Scaled"],
    index=df.index
)

print(df_scaled)
```
