```python
column_name = 'YearsExperience'

Q1 = df[column_name].quantile(0.25)
Q3 = df[column_name].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

iqr_outliers = df[(df[column_name] < lower_bound) | (df[column_name] > upper_bound)]
clean_iqr = df[(df[column_name] >= lower_bound) & (df[column_name] <= upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:\n", iqr_outliers)
print("Outliers:\n", len(iqr_outliers))
print("Data Without Outliers:\n",clean_iqr)
print("Data Without Outliers:\n",len(clean_iqr)) 
```

```python
column_name = 'Salary'

Q1 = df[column_name].quantile(0.25)
Q3 = df[column_name].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

iqr_outliers = df[(df[column_name] < lower_bound) | (df[column_name] > upper_bound)]
clean_iqr = df[(df[column_name] >= lower_bound) & (df[column_name] <= upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:\n", iqr_outliers)
print("Outliers:\n", len(iqr_outliers))
print("Data Without Outliers:\n",clean_iqr)
print("Data Without Outliers:\n",len(clean_iqr)) 
```

```python
x=df[['YearsExperience']]
print(x)
```

```python
y=df[['Salary']]
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=12
)
```

```python
from sklearn.linear_model import LinearRegression
slr=LinearRegression()
slr.fit(x_train,y_train)
slr.predict(x_test)
```

```python
from sklearn.metrics import r2_score
y_pred=slr.predict(x_test)
r2=r2_score(y_test,y_pred)
print(r2)
```
